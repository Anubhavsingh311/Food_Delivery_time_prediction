```python
df_temp["Delivery_person_Age"] = pd.to_numeric(
    df_temp["Delivery_person_Age"],
    errors="coerce"
)

df_temp["Delivery_person_Ratings"] = pd.to_numeric(
    df_temp["Delivery_person_Ratings"],
    errors="coerce"
)

df_temp["multiple_deliveries"] = pd.to_numeric(
    df_temp["multiple_deliveries"],
    errors="coerce"
)
```

```python
df_temp["Delivery_person_Ratings"] = df_temp["Delivery_person_Ratings"].fillna(df_temp["Delivery_person_Ratings"].median())
df_temp["Delivery_person_Ratings"].isna().sum()
```

```python
df_temp["multiple_deliveries"] = df_temp["multiple_deliveries"].fillna(df_temp["multiple_deliveries"].mode()[0])
df_temp["multiple_deliveries"].isna().sum()
```

```python
df_temp["Weatherconditions"] = df_temp["Weatherconditions"].fillna("Unknown")
df_temp["Road_traffic_density"] = df_temp["Road_traffic_density"].fillna("Unknown")
df_temp["Festival"] = df_temp["Festival"].fillna("Unknown")
df_temp["City"] = df_temp["City"].fillna("Unknown")
df_temp.isna().sum()
```

```python
#Feature Engineering
#1) Calculating distance using Haversine formula
def haversine_distance(lat1, lon1, lat2, lon2):
    R = 6371 

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    )

    c = 2 * np.arcsin(np.sqrt(a))

    return R * c
```

```python
df_temp["distance_km"] = haversine_distance(
    df_temp["Restaurant_latitude"],
    df_temp["Restaurant_longitude"],
    df_temp["Delivery_location_latitude"],
    df_temp["Delivery_location_longitude"]
)
```

```python
Q1 = df_temp["distance_km"].quantile(0.25)
Q3 = df_temp["distance_km"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_temp = df_temp[(df_temp["distance_km"] >= lower_bound) & (df_temp["distance_km"] <= upper_bound)].copy()
df_temp["distance_km"].describe()
```
