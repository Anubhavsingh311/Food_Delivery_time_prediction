```python
#Feature Engineering
#1) Calculating distance using Haversine formula
def haversine_distance(lat1, lon1, lat2, lon2):
    R = 6371 

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    )

    c = 2 * np.arcsin(np.sqrt(a))

    return R * c
```

```python
df_temp["distance_km"] = haversine_distance(
    df_temp["Restaurant_latitude"],
    df_temp["Restaurant_longitude"],
    df_temp["Delivery_location_latitude"],
    df_temp["Delivery_location_longitude"]
)
```

```python
#Encoding and training
# Separate input features and target

X = df.drop(columns=["Time_taken(min)"])
y = df["Time_taken(min)"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Columns with a natural order
ordinal_cols = [
    "Road_traffic_density",
    "Time_Period",
    "Pickup_Time_Period"
]

# Columns without a natural order
onehot_cols = [
    "Weatherconditions",
    "Type_of_order",
    "Type_of_vehicle",
    "Festival",
    "City"
]
```

```python
for col in ordinal_cols:
    print(f"\n--- {col} ---")
    print(X[col].unique())
```

```python
#Breaking data for training and testing
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
#Making encoder??
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.compose import ColumnTransformer

ordinal_categories = [
    ["Low", "Medium", "High", "Jam", "Unknown"],       # Road_traffic_density
    ["Morning", "Afternoon", "Evening", "Night", "Unknown"],  # Time_Period
    ["Morning", "Afternoon", "Evening", "Night"]        # Pickup_Time_Period
]

ordinal_encoder = OrdinalEncoder(
    categories=ordinal_categories,
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

onehot_encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)
```

```python
#Transformer
preprocessor = ColumnTransformer(
    transformers=[
        ("ordinal", ordinal_encoder, ordinal_cols),
        ("onehot", onehot_encoder, onehot_cols)
    ],
    remainder="passthrough"
)
```

```python
#Fitting
X_train_encoded = preprocessor.fit_transform(X_train)

# Transform test data using the already-fitted preprocessor
X_test_encoded = preprocessor.transform(X_test)

print("Encoded X_train shape:", X_train_encoded.shape)
print("Encoded X_test shape :", X_test_encoded.shape)
```

```python
# Linear Regression

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

linear_model = LinearRegression()

# Train
linear_model.fit(X_train_encoded, y_train)

# Predict
y_pred_linear = linear_model.predict(X_test_encoded)

# Evaluation
mae_linear = mean_absolute_error(y_test, y_pred_linear)
rmse_linear = np.sqrt(mean_squared_error(y_test, y_pred_linear))
r2_linear = r2_score(y_test, y_pred_linear)

print("Linear Regression Performance")
print("--------------------------------")
print(f"MAE  : {mae_linear:.2f} minutes")
print(f"RMSE : {rmse_linear:.2f} minutes")
print(f"R²   : {r2_linear:.4f}")
print(f"Accuracy (R²): {r2_linear * 100:.2f}%")
```
